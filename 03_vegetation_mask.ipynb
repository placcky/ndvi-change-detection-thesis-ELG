# 🌿 Vegetation Threshold Masking

## Purpose

This notebook applies an NDVI threshold mask to seasonal NDVI mosaics in order to isolate vegetation-covered areas.

Pixels below the selected threshold are converted to NoData values, while vegetation pixels are preserved.

---

## Input

Expected input rasters:

```text
NDVI_spring_2020.tif
NDVI_summer_2020.tif
```

Input folder:

```text
NDVI_seasonal_mosaic
```

---

## Threshold Used

```text
NDVI >= 0.5
```

This threshold preserves medium and high vegetation values while removing:

- bare soil
- urban surfaces
- weak vegetation
- non-vegetated areas

---

## Output

Example outputs:

```text
NDVI_spring_2020_veg.tif
NDVI_summer_2020_veg.tif
```


## 1️⃣ Import Libraries

Import ArcPy, OS utilities, and Spatial Analyst tools required for raster masking.


````python
import arcpy
import os
from arcpy.sa import *

# Enable Spatial Analyst extension
arcpy.CheckOutExtension("Spatial")

# Allow overwriting outputs
arcpy.env.overwriteOutput = True
````

## 2️⃣ Define Input and Output Folders

Specify the seasonal mosaic input folder and the output folder for vegetation-only rasters.


````python
# Update these paths before running the notebook
input_folder = r"path\to\input\NDVI_seasonal_mosaic"
output_folder = r"path\to\output\NDVI_filtered"

# Create output folder if it does not exist
if not os.path.exists(output_folder):
    os.makedirs(output_folder)
````

## 3️⃣ Apply Vegetation Threshold Mask

This step:

- Loads each NDVI raster
- Applies the vegetation threshold
- Converts low NDVI values to NoData
- Preserves vegetation-covered pixels
- Exports filtered rasters


````python
# --------------------------------------------------
# Apply vegetation threshold masking
# --------------------------------------------------

for f in os.listdir(input_folder):

    # Process only raster files
    if f.lower().endswith((".tif", ".tiff")):

        input_path = os.path.join(input_folder, f)

        print("Processing:", f)

        # Load raster
        ndvi = Raster(input_path)

        # Keep only vegetation pixels
        vegetation = SetNull(ndvi < 0.5, ndvi)

        # Generate output filename
        out_name = f.replace(".tif", "_veg.tif")
        out_path = os.path.join(output_folder, out_name)

        # Save output raster
        vegetation.save(out_path)

        print("✔ Saved:", out_name)

print("\nVegetation masking completed.")
````
